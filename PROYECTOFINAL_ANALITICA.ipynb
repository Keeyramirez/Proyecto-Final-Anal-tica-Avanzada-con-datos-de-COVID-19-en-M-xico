## **Limpieza**

**Variables importantes**


1. EDAD
2. NEUMONIA
3. INTUBADO
4. UCI
5. DIABETES
6. HIPERTENSION
7. OBESIDAD
8. FALLECIO (derivada de FECHA_DEF)
9. SEXO

```python
# EDAD
df = df[(df["EDAD"] >= 0) & (df["EDAD"] <= 110)]

# variable FALLECIO
df["FALLECIO"] = df["FECHA_DEF"].apply(
    lambda x: 0 if x == "9999-99-99" else 1)

# Variables clinicas
vars_clinicas = [
    "NEUMONIA", "INTUBADO", "UCI",
    "DIABETES", "HIPERTENSION", "OBESIDAD"]

# eliminar solo 99
for col in vars_clinicas:
    df = df[df[col] != 99]

# Re-codificar:
# 1 = Si,  2 = No, 97 = No aplica, 98 = Se ignora = 0
mapeo = {1: 1, 2: 0, 97: 0, 98: 0}
for col in vars_clinicas:
    df[col] = df[col].map(mapeo)

# SEXO
df = df[df["SEXO"] != 99]
df["SEXO"] = df["SEXO"].map({1: 0, 2: 1})
```

## **NORMALIZACION**

```python
# el escalador Min-Max
scaler = MinMaxScaler()

# normalizar la variable edad
df["EDAD_NORMALIZADA"] = scaler.fit_transform(df[["EDAD"]])

df[["EDAD", "EDAD_NORMALIZADA"]].head()
```

## **Propuesta solucion**

```python
# Variables finales para clustering
features = [
    "EDAD_NORMALIZADA","NEUMONIA",
    "INTUBADO", "UCI", "DIABETES",
    "HIPERTENSION", "OBESIDAD", "FALLECIO"]

X = df[features]
```

**K-means**

```python
from sklearn.cluster import KMeans

# Definir número de clusters
kmeans = KMeans(n_clusters=3, random_state=42)

# Ajustar modelo
df["CLUSTER_KMEANS"] = kmeans.fit_predict(X)
```

```python
# Tamaño de cada cluster
df["CLUSTER_KMEANS"].value_counts()
```

```python
# Promedio de variables clínicas por cluster
df.groupby("CLUSTER_KMEANS")[features].mean()
```

```python
cluster_colors = {
    0: '#FFD700',  # Dorado
    1: '#ff7f0e',  # Naranja
    2: '#FF0000'   # Rojo
    }
```

```python
cluster_counts = df['CLUSTER_KMEANS'].value_counts().sort_index()

plt.figure()
plt.bar(
    cluster_counts.index,
    cluster_counts.values,
    color=[cluster_colors[c] for c in cluster_counts.index]
)
plt.title('Tamaño de los clusters (K-means)')
plt.xlabel('Cluster')
plt.ylabel('Número de pacientes')
plt.show()
```

```python
variables = [
    "EDAD", "NEUMONIA", "INTUBADO", "UCI",
    "DIABETES","HIPERTENSION", "OBESIDAD", "FALLECIO"]

X = df[variables].copy()

# normalizacion
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# preparar datos para Fuzzy C-Means
X_fuzzy = X_scaled.T

# alicar C-means
n_clusters = 3

cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    X_fuzzy,
    c=n_clusters,
    m=2,
    error=0.005,
    maxiter=1000,
    seed=42)

# asignar cluster
df["CLUSTER_CMEANS"] = np.argmax(u, axis=0)

# resumen por cluster
resumen_cmeans = df.groupby("CLUSTER_CMEANS")[variables].mean()
conteo_cmeans = df["CLUSTER_CMEANS"].value_counts()

print("Conteo de pacientes por cluster:")
print(conteo_cmeans)

print("\nPerfil promedio por cluster:")
print(resumen_cmeans)

# coeficiente de particion difusa
print("\nFuzzy Partition Coefficient (FPC):", fpc)
```
